```python
import numpy as np
import math 
import time
import random

# Generation0(初代親)生成
n = 11  # 個体のバイナリデータの桁数
x = 10  # 親の数（ランダムに選ぶ数）
nisinsu = []
nisinsu_random = []

def all_binaries(n):
    for i in range(2**n):
        bits = []
        num = i
        for j in range(n):
            bits.insert(0, num % 2)
            num //= 2
        nisinsu.append(bits)
    return nisinsu

binaries = all_binaries(n)
nisinsu_random = random.sample(nisinsu, x)

# 各親のfitnessを計算してリスト化
def calc_fitness(individual):
    A, B = 0, 0
    for j in range(n):
        if individual[j] == 1:
            A += math.sqrt(j + 1)
        else:
            B += math.sqrt(j + 1)
    return abs(A - B)

fitness_list = []
for i in range(len(nisinsu_random)):
    f = calc_fitness(nisinsu_random[i])
    fitness_list.append(f)
    print(i, ")", nisinsu_random[i], ", fitness:", f)

# -------------------------------
# トーナメント選択関数
# -------------------------------
def tournament_selection(population, fitness_list, k=2):
    """k個体をランダムに選び、fitnessが最小の個体を返す"""
    candidates = random.sample(list(enumerate(population)), k)
    best = min(candidates, key=lambda x: fitness_list[x[0]])
    return best[1]  # 個体データを返す

# Generation更新
def gene_update():
    gene_best = 20  # 初期値（フロイド版）
    elite = []

    global nisinsu_random, fitness_list

    for u in range(10):

        # -------------------------------
        # トーナメント選択による親選び
        # -------------------------------
        parent1 = tournament_selection(nisinsu_random, fitness_list)
        parent2 = tournament_selection(nisinsu_random, fitness_list)
        parents = [parent1, parent2]

        # 親のインデックスを取得
        index_all = [nisinsu_random.index(parent1), nisinsu_random.index(parent2)]

        # -------------------------------
        # 交叉
        # -------------------------------
        s = random.randint(1, 10)
        offspring_kousa = []
        for g in range(2):
            if g == 0:
                for h in range(s):
                    offspring_kousa.append(parents[g][h])
            elif g == 1:
                for h in range(s, n):
                    offspring_kousa.append(parents[g][h])

        # -------------------------------
        # 突然変異
        # -------------------------------
        mutation_counter = 0
        for j in range(n):
            mutation_rate = random.random()
            if mutation_rate <= 0.05:
                mutation_counter += 1
                offspring_kousa[j] = 1 - offspring_kousa[j]  # 0↔1反転

        # 個体を更新
        nisinsu_random[u] = offspring_kousa

        # -------------------------------
        # 評価関数（フロイド版）
        # -------------------------------
        fitness_gen1 = calc_fitness(offspring_kousa)
        if fitness_gen1 <= gene_best:
            gene_best = fitness_gen1
            elite = offspring_kousa.copy()

        # -------------------------------
        # 表示
        # -------------------------------
        if mutation_counter > 0:
            print(f"{u}）（{index_all[0]},{index_all[1]}）,交叉点位置：{s},子供：{offspring_kousa},fitness:{fitness_gen1}（突然変異発生）")
        else:
            print(f"{u}）（{index_all[0]},{index_all[1]}）,交叉点位置：{s},子供：{offspring_kousa},fitness:{fitness_gen1}")

    print(f"エリート個体は{elite}で、評価値は{gene_best}です")

    # エリートを最後の個体に残す
    nisinsu_random[-1] = elite

    # fitnessリストを更新
    fitness_list = [calc_fitness(ind) for ind in nisinsu_random]

    print(nisinsu_random)


# -------------------------------
# 進化ループ
# -------------------------------
for i in range(50):
    print("世代", i + 1)
    gene_update()
```
